```python
def even_or_odd(num):
    if num % 2 == 0:
        return "even number"
    else: 
        return "odd number"
```

```python
even_or_odd(22)
```

```python
def greet(name="Aryan"):
    return f"hello {name}"
```

```python
def print_numbers(*args):
    for numbers in args:
        print(numbers)
```

```python
print_numbers(1,2,3,4,5,6,"Aryan")
```

```python
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")
```

```python
print_details(name="Aryan", age=32, country="India")
```

```python
def print_details(*args, **kwargs):
    for numbers in args:
        print(numbers)
    for key, value in kwargs.items():
        print(f"{key} : {value}")
```

```python
print_details(1,2,3,4,5,6,"Aryan")
```

```python
print_details(1, 2, 3, name="Aryan", age=21)
```
