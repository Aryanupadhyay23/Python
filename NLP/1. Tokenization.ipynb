# Basic NLP Concepts: Corpus, Documents, Tokens, and Vocabulary

## 1. Corpus

A **corpus** is a collection of text data. It can be a paragraph, multiple paragraphs, or even a large dataset of text.

**Example:**

> My name is Aryan and I have interest in learning Artificial Intelligence and Machine Learning. I am a student.

The entire text above is called a **corpus**.

---

## 2. Documents (Sentences)

A **document** refers to a smaller unit within a corpus. In many cases, each **sentence** in a paragraph is treated as a document.

**Example:**
From the corpus:

* Document 1: My name is Aryan and I have interest in learning Artificial Intelligence and Machine Learning.
* Document 2: I am a student.

---

## 3. Tokenization

**Tokenization** is the process of breaking text into smaller units called **tokens**.

Tokens can be:

* Sentences (Sentence Tokenization)
* Words (Word Tokenization)

### (a) Sentence Tokenization

Breaking a paragraph into sentences (documents).

**Example:**

```
["My name is Aryan and I have interest in learning Artificial Intelligence and Machine Learning.",
 "I am a student."]
```

---

### (b) Word Tokenization

Breaking a sentence into individual words.

**Example:**

```
["my", "name", "is", "aryan", "and", "i", "have", "interest", "in", "learning", "artificial", "intelligence", "and", "machine", "learning"]
```

So:

* A **sentence can be a token**
* A **word can also be a token**

---

## 4. Words vs Vocabulary

### Words

All the words present in the corpus (including repetition).

### Vocabulary

The set of **unique words** present in the corpus.

---

## 5. Example to Understand Vocabulary

**Paragraph (Corpus):**

> I like to drink apple juice. My friend likes mango juice.

### Step 1: Sentence Tokenization

```
["I like to drink apple juice.",
 "My friend likes mango juice."]
```

### Step 2: Word Tokenization

Sentence 1:

```
["i", "like", "to", "drink", "apple", "juice"]
```

Sentence 2:

```
["my", "friend", "likes", "mango", "juice"]
```

### Total Words

Total number of words = **11**

### Unique Words (Vocabulary)

```
["i", "like", "to", "drink", "apple", "juice", "my", "friend", "likes", "mango"]
```

Total unique words = **10** → This is the **vocabulary**

---

* **Corpus** → Entire text data
* **Documents** → Sentences inside the corpus
* **Tokenization** → Process of breaking text into tokens
* **Tokens** → Can be sentences or words
* **Words** → All words (including duplicates)
* **Vocabulary** → Unique words in the corpus

---

````python
corpus = """My name is Aryan. 
I'm Learning Artificial Intelligence and Machine Learning.
"""
````

````python
print(corpus)
````

## paragraphs -> sentences

````python
from nltk.tokenize import sent_tokenize
````

````python
documents = sent_tokenize(corpus)
````

````python
type(documents)
````

````python
documents
````

````python
for sentence in documents:
    print(sentence)
````

## sentences -> words
## paragraph -> words

````python
from nltk.tokenize import word_tokenize
````

````python
word_tokenize(corpus)
````

````python
for sentence in documents:
    print(word_tokenize(sentence))
````

````python
from nltk.tokenize import wordpunct_tokenize
````

````python
wordpunct_tokenize(corpus)
````

````python
from nltk.tokenize import TreebankWordTokenizer
````

````python
tokenizer = TreebankWordTokenizer()
````

````python
tokenizer.tokenize(corpus)
````
