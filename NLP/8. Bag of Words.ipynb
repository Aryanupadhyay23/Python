## Bag of Words (BoW)

**Bag of Words** is a text representation technique used in NLP where a sentence or document is represented as a **vector of word frequencies**, ignoring grammar and word order.

---

## How it Works (Step-by-Step)

1. **Collect Text Data**
   Gather all sentences/documents (corpus)

2. **Preprocessing**

   * Convert to lowercase
   * Remove punctuation
   * Remove stopwords (optional)
   * Apply stemming/lemmatization (optional)

3. **Tokenization**
   Split text into individual words (tokens)

4. **Build Vocabulary**
   Create a list of **unique words** from the corpus

5. **Vector Representation**
   Count how many times each word appears in a sentence/document

6. **Create Matrix**
   Rows → documents
   Columns → vocabulary
   Values → word frequency

---

## Example

Sentence 1: "I love machine learning"
Sentence 2: "I love deep learning"

Vocabulary:

* I, love, machine, learning, deep

BoW Representation:

| Sentence | I | love | machine | learning | deep |
| -------- | - | ---- | ------- | -------- | ---- |
| S1       | 1 | 1    | 1       | 1        | 0    |
| S2       | 1 | 1    | 0       | 1        | 1    |

---

## Why Use Bag of Words?

* Converts **text → numerical format**
* Easy to implement
* Works well for **basic NLP tasks** like:

  * Text classification
  * Spam detection
  * Sentiment analysis

---

## Advantages

* Simple and fast
* Easy to understand
* Effective for small datasets
* No complex computation required

---

## Disadvantages

* **Ignores word order** (context is lost)
* **No semantic meaning captured**
* High **dimensionality** with large vocabulary
* Produces **sparse matrix**
* Cannot handle **Out of Vocabulary (OOV)** words
* Does not capture relationships between words

---


* BoW focuses only on **word frequency**, not meaning or position
* Good starting point for NLP, but limited for complex understanding


## Types of Bag of Words (BoW)

Bag of Words can be implemented in different ways based on how we represent word values.

---

## 1. Binary Bag of Words

In **Binary BoW**, we only check whether a word is present or not in a document.

* If word is present → **1**
* If word is not present → **0**

### Example

Sentence: "I love machine learning learning"

Vocabulary:

* I, love, machine, learning

Binary Representation:

| I | love | machine | learning |
| - | ---- | ------- | -------- |
| 1 | 1    | 1       | 1        |

-> Even though "learning" appears twice, it is marked as **1** (only presence matters)

---

## 2. Normal Bag of Words (Count BoW)

In **Normal BoW**, we count how many times each word appears in the document.

* Value = **frequency of word**

### Example

Sentence: "I love machine learning learning"

Vocabulary:

* I, love, machine, learning

Count Representation:

| I | love | machine | learning |
| - | ---- | ------- | -------- |
| 1 | 1    | 1       | 2        |

-> Here, "learning" appears twice, so value = **2**

---

## Key Difference

| Feature            | Binary BoW  | Normal BoW                |
| ------------------ | ----------- | ------------------------- |
| Value Type         | 0 or 1      | Frequency count           |
| Focus              | Presence    | Frequency                 |
| Handles repetition | No          | Yes                       |
| Simplicity         | More simple | Slightly more informative |

---


* **Binary BoW** → Checks only presence of words
* **Normal BoW** → Counts frequency of words
* Both ignore **word order and semantics**
* Choice depends on the problem and model requirement


```python
import nltk
import re
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer
```

```python
nltk.download('stopwords')
```

```python
corpus = [
    "This product is really amazing and I am very happy with its performance",
    "I did not like the product it was bad and disappointing",
    "The quality is good but the delivery was late",
    "Absolutely fantastic experience I love this product"
]
```

```python
stop_words = set(stopwords.words('english'))
ps = PorterStemmer()
```

```python
processed_corpus = []

for sentence in corpus:
    # Lowercase
    sentence = sentence.lower()
    
    # Remove special characters
    sentence = re.sub('[^a-z]', ' ', sentence)
    
    # Tokenization
    words = sentence.split()
    
    # Remove stopwords + apply stemming
    words = [ps.stem(word) for word in words if word not in stop_words]
    
    # Join back to sentence
    sentence = ' '.join(words)
    
    processed_corpus.append(sentence)

processed_corpus
```

```python
cv = CountVectorizer()
X = cv.fit_transform(processed_corpus)
```

```python
print(cv.get_feature_names_out())
```

```python
print(X.toarray())
```

```python
bow_df = pd.DataFrame(X.toarray(), columns=cv.get_feature_names_out())
bow_df
```

## N-grams in NLP

**N-grams** are continuous sequences of **N words** taken from a given text.
They help capture **context and word order**, which is missing in simple Bag of Words.

---

## What are N-grams?

* If **N = 1** → Unigram (single words)
* If **N = 2** → Bigram (pair of words)
* If **N = 3** → Trigram (three-word sequence)

---

## Example

Sentence:

> I love machine learning

### Unigrams (N = 1)

* I
* love
* machine
* learning

### Bigrams (N = 2)

* I love
* love machine
* machine learning

### Trigrams (N = 3)

* I love machine
* love machine learning

---

## Why Use N-grams?

* Captures **word order and context**
* Helps understand **phrases instead of single words**
* Improves performance in:

  * Sentiment analysis
  * Text classification
  * Language modeling

Example:

* "not good" vs "good" → meaning changes
* Bigrams help capture this difference

---

## How N-grams Work

1. Take a sentence
2. Choose value of **N**
3. Slide a window of size **N** across the sentence
4. Extract word sequences

---

## Advantages

* Captures **context and relationships between words**
* More informative than single words
* Helps detect phrases

---

## Disadvantages

* Increases **dimensionality**
* Creates **sparse matrix**
* Requires more data for training
* Still limited in capturing deep semantics

---


* N-grams = sequence of N words
* Helps capture **context and order**
* Better than Bag of Words but more complex
* Common types: **Unigram, Bigram, Trigram**


```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
corpus = [
    "I love machine learning",
    "I love deep learning",
    "Machine learning is amazing"
]
```

```python
cv_unigram = CountVectorizer(ngram_range=(1,1))
X_uni = cv_unigram.fit_transform(corpus)

print(cv_unigram.get_feature_names_out())
print(X_uni.toarray())
```

```python
cv_bigram = CountVectorizer(ngram_range=(2,2))
X_bi = cv_bigram.fit_transform(corpus)

print(cv_bigram.get_feature_names_out())
print(X_bi.toarray())
```

```python
cv_trigram = CountVectorizer(ngram_range=(3,3))
X_tri = cv_trigram.fit_transform(corpus)

print(cv_trigram.get_feature_names_out())
print(X_tri.toarray())
```

```python
cv_combined = CountVectorizer(ngram_range=(1,2))
X_comb = cv_combined.fit_transform(corpus)

print(cv_combined.get_feature_names_out())
print(X_comb.toarray())
```

```python
df = pd.DataFrame(X_comb.toarray(), columns=cv_combined.get_feature_names_out())
df
```
