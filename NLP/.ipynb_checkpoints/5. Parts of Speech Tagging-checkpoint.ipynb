## Parts of Speech (POS Tags)

Parts of Speech (POS) represent the grammatical role of words in a sentence. In NLP, POS tagging helps in understanding sentence structure and meaning.

---

### Common POS Tags

| Tag  | Meaning                               | Example             |
| ---- | ------------------------------------- | ------------------- |
| CC   | Coordinating Conjunction              | and, but, or        |
| CD   | Cardinal Digit                        | 1, 100              |
| DT   | Determiner                            | the, a, an          |
| EX   | Existential There                     | there is, there are |
| FW   | Foreign Word                          | bonjour             |
| IN   | Preposition/Subordinating Conjunction | in, on, because     |
| JJ   | Adjective                             | big, happy          |
| JJR  | Adjective (Comparative)               | bigger              |
| JJS  | Adjective (Superlative)               | biggest             |
| LS   | List Marker                           | 1), a)              |
| MD   | Modal                                 | can, will, should   |
| NN   | Noun (Singular)                       | desk                |
| NNS  | Noun (Plural)                         | desks               |
| NNP  | Proper Noun (Singular)                | Aryan, India        |
| NNPS | Proper Noun (Plural)                  | Indians             |
| PDT  | Predeterminer                         | all, both           |
| POS  | Possessive Ending                     | ’s                  |
| PRP  | Personal Pronoun                      | I, he, she          |
| PRP$ | Possessive Pronoun                    | my, his, her        |
| RB   | Adverb                                | quickly, very       |
| RBR  | Adverb (Comparative)                  | faster              |
| RBS  | Adverb (Superlative)                  | fastest             |
| VB   | Verb (Base Form)                      | run                 |
| VBD  | Verb (Past Tense)                     | ran                 |
| VBG  | Verb (Gerund)                         | running             |
| VBN  | Verb (Past Participle)                | eaten               |
| VBP  | Verb (Present)                        | run                 |
| VBZ  | Verb (3rd person singular)            | runs                |
| WDT  | Wh-determiner                         | which, that         |
| WP   | Wh-pronoun                            | who, what           |
| WRB  | Wh-adverb                             | where, when         |

---

### Example

Sentence:

> I am learning Natural Language Processing

POS Tags:

* I → PRP
* am → VBP
* learning → VBG
* Natural → JJ
* Language → NN
* Processing → NN

---

* POS tagging helps in **understanding grammar**
* Useful in **lemmatization, parsing, and NLP models**
* Improves **context understanding** in text processing


```python
paragraph = """Machine learning is a powerful field of artificial intelligence that enables systems to learn patterns from data and make predictions or decisions without being explicitly programmed. 
It is widely used in real-world applications such as recommendation systems, fraud detection, healthcare diagnostics, and self-driving cars. 
By leveraging algorithms and statistical techniques, machine learning models improve their performance over time as they are exposed to more data. 
Understanding the fundamentals of data preprocessing, model selection, and evaluation is essential for building effective and reliable machine learning solutions."""
```

```python
import nltk
from nltk.corpus import stopwords
sentences = nltk.sent_tokenize(paragraph)
```

```python
print(sentences)
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [word for word in words if word not in set(stopwords.words('english'))]
    pos_tag = nltk.pos_tag(words)
    print(pos_tag)
```

```python
"Taj Mahal is a beautiful Monument".split()
```

```python
print(nltk.pos_tag("Taj Mahal is a beautiful Monument".split()))
```
