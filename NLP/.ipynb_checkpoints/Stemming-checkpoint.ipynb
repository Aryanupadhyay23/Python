# Stemming in NLP

## Problem in Text Data

In many NLP tasks like **sentiment analysis (positive/negative review classification)**, text data contains different forms of the same word.

**Example:**

* eat, eating, eaten
* go, going, goes, gone

Although these words look different, they represent the **same core meaning**.

---

## Issue with Multiple Word Forms

If we treat all these variations as separate words:

* It **increases dimensionality** (more features)
* It creates **redundancy**
* It may not improve model performance, since all forms convey similar meaning

For example:

* "eat", "eating", "eaten" → same intent in sentiment
* Model does not need to treat them as completely different features

---

## What is Stemming?

**Stemming** is a text preprocessing technique used to reduce words to their **root (stem) form**.

The root word is called the **stem**, which may or may not be a valid English word.

---

## Example of Stemming

| Original Words      | Stemmed Word |
| ------------------- | ------------ |
| eating, eaten, eats | eat          |
| going, goes, gone   | go           |

---

## Why Stemming is Important

* Reduces **vocabulary size**
* Decreases **dimensionality**
* Removes **redundant features**
* Improves **model efficiency**
* Helps focus on the **core meaning** of words

---

## Key Insight

Instead of treating:

* "eating", "eaten", "eat" as different words

We convert them into:

* "eat" → a single feature

This simplifies the problem and makes the model more efficient.

---

* Text data contains multiple forms of the same word
* These variations increase unnecessary complexity
* **Stemming reduces words to their root form**
* Helps in efficient text representation and better model performance

---

## Disadvantages of Stemming

- Over-stemming: Different words may be reduced to the same stem incorrectly
- Loss of meaning: Stems are not always real words

Examples:

* history → histori
* congratulations → congratul
* studies → studi
* Can sometimes produce incorrect or confusing words
* Not context-aware (does not understand meaning, only rules)


## PorterStemmer

````python
from nltk.stem import PorterStemmer
````

````python
stemming = PorterStemmer()
````

````python
words = [
    "eating", "eaten", "eats",
    "running", "runs", "ran",
    "playing", "played", "plays",
    "studies", "studying", "studied",
    "fairly", "easily", "happiness",
    "goes", "going", "gone"
]
````

````python
for word in words:
    print(word + " -----> " + stemming.stem(word))
````

````python
stemming.stem("congratulations")
````

````python
stemming.stem("sitting")
````

### RegexpStemmer Class

In NLTK, the **RegexpStemmer** is used to perform stemming using **regular expressions (regex)**.

It allows us to define our own rules to remove suffixes or prefixes from words.

---

### How it Works

* Takes a **regex pattern**
* Matches that pattern in words
* Removes the matched part

---

### Example

```
from nltk.stem import RegexpStemmer

# Remove 'ing', 'ed', 's' from end of words
stemmer = RegexpStemmer('ing$|ed$|s$', min=4)

words = ["playing", "played", "plays", "player"]

for word in words:
    print(word, "->", stemmer.stem(word))
```

---

### Output

```
playing -> play
played  -> play
plays   -> play
player  -> player
```

---

### Explanation

* `ing$` → removes words ending with "ing"
* `ed$` → removes words ending with "ed"
* `s$` → removes words ending with "s"
* `$` → ensures match at end of word
* `min=4` → apply stemming only if word length ≥ 4

---

### Key Points

* Rule-based stemming using regex
* Fully customizable
* Reduces vocabulary size
* May not always give correct English words

---

* RegexpStemmer uses **custom regex rules**
* Helps remove suffixes/prefixes
* Useful for controlled and simple stemming tasks


````python
from nltk.stem import RegexpStemmer
````

````python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
````

````python
reg_stemmer.stem('eating')
````

### Snowball Stemmer

The **Snowball Stemmer** is a stemming algorithm in NLP. It is also known as the **Porter2 Stemmer**, as it is an improved version of the Porter Stemmer.

It was developed to fix some limitations of the original Porter Stemmer and provide more consistent and accurate stemming results.

---

### Key Features

* Improved version of Porter Stemmer
* More accurate and consistent stemming
* Supports multiple languages (English, French, German, etc.)
* Less aggressive compared to some other stemmers

---

### Example

```
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")

words = ["running", "runs", "ran", "easily", "fairly"]

for word in words:
    print(word, "->", stemmer.stem(word))
```

---

### Output

```
running -> run  
runs    -> run  
ran     -> ran  
easily  -> easili  
fairly  -> fairli  
```

---

### Explanation

* Converts different forms of words into a common root
* "running" and "runs" → "run"
* Produces stems that may not always be valid English words

---

* Snowball Stemmer = Improved Porter Stemmer
* Provides better stemming results
* Supports multiple languages
* Commonly used in NLP preprocessing


````python
from nltk import SnowballStemmer
````

````python
snowballstemmer = SnowballStemmer('english')
````

````python
for word in words:
    print(word + " ----> " + snowballstemmer.stem(word))
````

````python
stemming.stem('fairly'), stemming.stem("sportingly")
````

````python
snowballstemmer.stem('fairly'), snowballstemmer.stem("sportingly")
````

````python
snowballstemmer.stem('going')
````

## Porter Stemmer vs Regexp Stemmer vs Snowball Stemmer

| Feature          | Porter Stemmer                 | Regexp Stemmer                        | Snowball Stemmer                 |
| ---------------- | ------------------------------ | ------------------------------------- | -------------------------------- |
| Type             | Rule-based (predefined rules)  | Custom rule-based (regex)             | Improved rule-based (Porter2)    |
| Approach         | Uses a set of linguistic rules | Uses user-defined regular expressions | Uses optimized and refined rules |
| Flexibility      | Low (fixed rules)              | High (fully customizable)             | Medium                           |
| Accuracy         | Moderate                       | Depends on regex (can be low/high)    | Higher than Porter               |
| Language Support | Mainly English                 | Any (depends on regex)                | Multiple languages               |
| Ease of Use      | Easy                           | Requires regex knowledge              | Easy                             |
| Output Quality   | Sometimes aggressive           | Depends on pattern                    | More consistent                  |
| Example          | running → run                  | running → run (if regex defined)      | running → run                    |
| Drawback         | Can produce incorrect stems    | Depends on user-defined rules         | Slightly slower than Porter      |
