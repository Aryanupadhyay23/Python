## Named Entity Recognition (NER) using NLTK

**Named Entity Recognition (NER)** is a technique in NLP used to identify and classify important entities in text into predefined categories such as:

* Person (e.g., Aryan, Elon Musk)
* Organization (e.g., Google, Microsoft)
* Location (e.g., India, New York)
* Date, Time, Money, etc.

---

## Why NER is Important

* Helps extract **useful information** from text
* Used in **chatbots, search engines, and information extraction**
* Improves understanding of **real-world entities**

---

## Steps to Perform NER in NLTK

1. Tokenization
2. POS Tagging
3. Named Entity Chunking

---

## Common Entity Labels in NLTK

| Label        | Meaning                      |
| ------------ | ---------------------------- |
| PERSON       | Names of people              |
| ORGANIZATION | Companies, institutions      |
| GPE          | Countries, cities, locations |
| LOCATION     | General locations            |
| DATE         | Dates                        |
| TIME         | Time expressions             |
| MONEY        | Monetary values              |
| PERCENT      | Percentage values            |

---

* NER identifies **real-world entities** in text
* Uses **POS tagging + chunking** in NLTK
* Output is a **tree structure**
* Useful for **information extraction and NLP applications**


```python
sentence = """The Eiffel Tower was built from 1887 to 1889 by French engineer 
Gustave Eiffel, whose company specialized in building metal frameworks and structures."""
```

```python
import nltk
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
!pip install svgling
```

```python
words = nltk.word_tokenize(sentence)
```

```python
tag_elements = nltk.pos_tag(words)
```

```python
nltk.ne_chunk(tag_elements).draw()
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag, ne_chunk

text = "Aryan lives in India and works at Google"

# Tokenization
tokens = word_tokenize(text)

# POS Tagging
pos_tags = pos_tag(tokens)

# Named Entity Recognition
ner = ne_chunk(pos_tags)

print(ner)
```
