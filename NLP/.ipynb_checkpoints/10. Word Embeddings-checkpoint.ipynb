<center><h1>Word Embeddings</h1></center>

## What are Word Embeddings?

In **Natural Language Processing (NLP)**, **word embeddings** are a technique used to represent words as **dense numerical vectors**.

These vectors are **real-valued** and capture the **meaning of words**, such that words with similar meanings are located **close to each other in vector space**.

---

## Why Word Embeddings?

Traditional methods like:

* Bag of Words
* One Hot Encoding

have limitations:

* No semantic meaning
* High dimensionality
* Sparse representation

Word embeddings solve these issues by:

* Capturing **semantic relationships**
* Reducing **dimensionality**
* Providing **dense representations**

---

## Key Idea

* Each word is represented as a **vector of numbers**
* Similar words → **similar vectors**
* Dissimilar words → **far apart vectors**

Example:

* king ≈ queen
* man ≈ woman

---

## Properties of Word Embeddings

* Dense (no unnecessary zeros)
* Low dimensional compared to one-hot encoding
* Capture **context and meaning**
* Enable mathematical relationships between words

---

## Example

Instead of representing:

* king → [0, 0, 1, 0, 0, ...] (One Hot)

We represent:

* king → [0.25, -0.10, 0.78, ...] (Embedding vector)

---

## Types of Word Embeddings

* **Word2Vec**

  * CBOW (Continuous Bag of Words)
  * Skip-gram

* **GloVe (Global Vectors)**

* **FastText**

---

## Advantages

* Captures **semantic meaning**
* Reduces dimensionality
* Improves performance of NLP models
* Handles similarity and relationships

---

## Disadvantages

* Requires training on large data
* Context-independent (same vector for a word in all contexts)
* May capture bias from training data

---


* Word embeddings represent words as **dense vectors**
* Words with similar meaning are **closer in vector space**
* More powerful than traditional text representation methods
* Widely used in modern NLP systems


## Word2Vec

**Word2Vec** is a neural network-based technique used in **Natural Language Processing (NLP)** to generate **word embeddings**. It converts words into **dense, low-dimensional vectors** such that words with similar meanings have similar vector representations.

It was introduced by **Tomas Mikolov (Google)** and is one of the most important methods for learning semantic representations of text.

---

## Core Idea

Word2Vec is based on the **distributional hypothesis**:

> "Words that appear in similar contexts tend to have similar meanings."

Instead of manually defining features, Word2Vec **learns features automatically** from large text data.

---

## How Word2Vec Works

* Input: Large corpus of text
* Process:

  * Slides a window over text
  * Learns relationships between target word and context words
* Output: Dense vector representation for each word

---

## Key Concepts

### 1. Context Window

* Defines how many surrounding words are considered
* Example (window size = 2):

Sentence:

> I love machine learning

For word **machine**, context = [love, I, learning]

---

### 2. Vector Representation

* Each word is mapped to a vector of fixed size (e.g., 100, 300 dimensions)
* These vectors capture semantic relationships

---

### 3. Training Objective

* Learn word vectors such that:

  * Similar context → similar vectors
  * Dissimilar context → different vectors

---

## Architectures of Word2Vec

### 1. CBOW (Continuous Bag of Words)

* Predicts the **target word** from surrounding context words
* Faster to train
* Works well for **small datasets**

**Example:**

> Context: "I love ___ learning"
> Target: "machine"

---

### 2. Skip-gram

* Predicts **context words** from a target word
* Slower but more accurate
* Works well for **large datasets**

**Example:**

> Target: "machine"
> Context: "I", "love", "learning"

---

## Training Mechanism (Simplified)

* Uses a shallow neural network
* Input layer → Hidden layer → Output layer
* Learns weights that become **word embeddings**

---

## Important Techniques

### 1. Negative Sampling

* Improves efficiency
* Instead of updating all words, updates only a few negative samples

### 2. Hierarchical Softmax

* Alternative to softmax
* Reduces computational cost

---

## Semantic Relationships

Word2Vec captures relationships using vector arithmetic:

* king - man + woman ≈ queen
* paris - france + italy ≈ rome

This shows it understands **analogies and relationships**.

---

## Advantages

* Captures **semantic and syntactic meaning**
* Dense and low-dimensional representation
* Efficient training
* Useful for downstream tasks:

  * Text classification
  * Sentiment analysis
  * Machine translation

---

## Disadvantages

* Context-independent (same vector for all meanings of a word)
* Needs large corpus for good performance
* Cannot handle unseen words (OOV problem)
* May learn biases from data

---

## Applications

* Search engines
* Chatbots
* Recommendation systems
* Text similarity
* Named Entity Recognition

---


* Word2Vec is a **neural embedding technique**
* Learns word meaning from **context**
* Two models: **CBOW and Skip-gram**
* Produces **dense vectors** that capture relationships
* Fundamental building block in modern NLP


# CBOW (Continuous Bag of Words)

## 1. Overview
**CBOW (Continuous Bag of Words)** is one of the two primary architectures of **Word2Vec** used to learn word embeddings (the other being Skip-gram). 

It works by **predicting a target word using its surrounding context words**. It is based on the distributional hypothesis: the idea that words appearing in similar contexts share similar meanings.

---

## 2. Intuition Behind CBOW
The core idea is simple: **A word can be understood by the company it keeps.** If we know the surrounding words, we can accurately guess the missing word.

### Example A: Fill in the Blank
> "I love \_\_\_ learning"

* **Context words:** I, love, learning
* **Target word:** machine

CBOW learns to predict **"machine"** using those context words.

### Example B: Contextual Similarity
* "I love machine learning"
* "I enjoy deep learning"

Here, words like *love* and *enjoy*, as well as *machine learning* and *deep learning*, appear in similar contexts. Because the network trains on these similar contexts to predict similar targets, their final word embeddings become mathematically similar.

---

## 3. How CBOW Works (The Process)
1. Take a sentence from your corpus.
2. Select a **target word**.
3. Identify surrounding **context words** based on a predefined **window size**.
4. Use context words as multiple inputs.
5. Predict the single target word.

### Example with Window Size = 2
Sentence: > *I love machine learning*

For target word **machine**:
* **Context:** [I, love, learning]
* **Training pair:** Input $\rightarrow$ [I, love, learning] | Output $\rightarrow$ machine

---

## 4. Technical Setup & One-Hot Encoding
To understand the math, let's look at a simpler setup with a **Window Size = 1**.

Sentence: > *the cat sits on the mat*

* **Vocabulary:** [the, cat, sits, on, mat]
* **Vocabulary size:** $V = 5$

Each word is represented using **one-hot encoding** (an array of mostly zeros with a single '1' at the word's index). 

For target word **"sits"**:
* Context = [cat, on]
* Target = sits

---

## 5. Model Architecture
CBOW uses a **shallow neural network** consisting of three layers. It uses **multiple input words to predict one output word**.

### 1. Input Layer
* Takes context words as one-hot vectors.
* There are multiple inputs (one per context word in the window).

### 2. Hidden Layer
* Size = $N$ (this is your desired embedding dimension).
* **No activation function** is used here.
* It simply computes the **average** of the context word embeddings.

### 3. Output Layer
* Size = $V$ (the total vocabulary size).
* Uses a **softmax** activation function to predict the probabilities of each word in the vocabulary being the target word.

---

## 6. Weight Matrices
The network relies on two weight matrices:
* **Input-to-Hidden matrix:** $W \in \mathbb{R}^{V \times N}$
* **Hidden-to-Output matrix:** $W' \in \mathbb{R}^{N \times V}$

**Important:** Matrix $W$ becomes the final word embeddings after training is complete. The $W'$ matrix is discarded.

---

## 7. Mathematical Formulation (Forward Propagation)

### Step 1: Hidden Representation ($h$)
The hidden layer averages the embeddings of the context words:

$$h = \frac{1}{C} \sum_{i=1}^{C} W^T x_i$$

* $C$ = number of context words
* $x_i$ = one-hot vector of the $i$-th context word

### Step 2: Output Scores ($u$)
The hidden representation is multiplied by the second weight matrix to get raw scores:

$$u = (W')^T h$$

### Step 3: Softmax ($\hat{y}$)
Convert raw scores into a probability distribution:

$$\hat{y}_j = \frac{e^{u_j}}{\sum_{k=1}^{V} e^{u_k}}$$

---

## 8. Loss Function & Backpropagation
To measure the error, CBOW uses Cross-Entropy Loss:

$$E = -\log(\hat{y}_{\text{target}})$$

Weights are updated using gradient descent to minimize this loss:

$$W_{\text{new}} = W_{\text{old}} - \eta \frac{\partial E}{\partial W}$$

*(where $\eta$ is the learning rate)*

---

## 9. Training Process
1. Slide the context window across the entire corpus.
2. Generate (context, target) training pairs.
3. Perform forward pass to get predictions.
4. Compute the loss $E$.
5. Backpropagate the error.
6. Update weights $W$ and $W'$.
7. Repeat for the entire corpus over multiple epochs.

### Common Hyperparameters
* **Embedding dimension ($N$):** 50, 100, 300
* **Window size ($C$):** 2–5
* **Learning rate ($\eta$)**

---

## 10. Advantages & Disadvantages

### Advantages
* **Speed:** Faster to train and computationally efficient compared to Skip-gram.
* **Data Efficiency:** Works exceptionally well with **smaller datasets**.
* **Robustness:** Averages the context, which smooths over noisy data and frequent/infrequent word variations.

### Disadvantages
* **Rare Words:** Does not perform well for rare words (Skip-gram is better here).
* **Ignores Word Order:** Because it averages the context embeddings (a "Bag of Words" concept), it loses grammatical positioning.
* **Nuance:** Less accurate than Skip-gram in capturing highly complex semantic relationships.

---

* CBOW predicts a target word using its **context**.
* Uses a **Many-to-One** architecture (multiple inputs $\rightarrow$ one output).
* Relies on **averaged embeddings** in the hidden layer.
* The matrix $W$ contains the final word embeddings.
* Result: Similar contexts yield similar vectors (e.g., *king* $\approx$ *queen*, *cat* $\approx$ *dog*).

# Skip-gram (Word2Vec)

## 1. Overview
**Skip-gram** is the second primary architecture of the **Word2Vec** model used to learn word embeddings (the other being CBOW). 

It works by flipping the CBOW approach on its head: instead of guessing a word based on its context, **Skip-gram predicts the surrounding context words using a single target word**. 

---

## 2. Intuition Behind Skip-gram
The core idea is: **A word dictates its environment.** If you give the model a specific word, it should be able to generate the words that typically surround it.

### Example A: Guess the Context
> "___ ___ machine ___"

* **Target word:** machine
* **Predicted Context words:** I, love, learning

Skip-gram learns that if the input is **"machine"**, there is a high probability that words like **"learning"**, **"learning"**, or **"state"** are nearby.

### Example B: Fine-Grained Semantics
Because Skip-gram evaluates a target word against *every* single context word individually, it is highly sensitive to nuances. It doesn't just blend the context together; it learns specific, distinct relationships (e.g., *apple* $\rightarrow$ *eat*, *apple* $\rightarrow$ *juice*, *apple* $\rightarrow$ *tree*).

---

## 3. How Skip-gram Works (The Process)
1. Take a sentence from your corpus.
2. Select a **target word**.
3. Identify surrounding **context words** based on a predefined **window size**.
4. Use the single target word as the input.
5. Predict the multiple context words.

### Example with Window Size = 2
Sentence: > *I love machine learning*

For target word **machine**:
* **Input:** machine
* **Context to predict:** [I, love, learning]
* **Training pairs:** 
  * Input $\rightarrow$ machine | Output $\rightarrow$ I
  * Input $\rightarrow$ machine | Output $\rightarrow$ love
  * Input $\rightarrow$ machine | Output $\rightarrow$ learning

*(Notice how Skip-gram breaks this into multiple distinct training steps for a single target word, which is why it learns such detailed representations).*

---

## 4. Technical Setup & One-Hot Encoding
Let's look at a simpler setup with a **Window Size = 1**.

Sentence: > *the cat sits on the mat*

* **Vocabulary:** [the, cat, sits, on, mat]
* **Vocabulary size:** $V = 5$

Each word is represented using **one-hot encoding**. 

For target word **"sits"**:
* Target (Input) = sits
* Context (Outputs) = [cat, on]

---

## 5. Model Architecture
Skip-gram uses a **shallow neural network** consisting of three layers. It uses **one input word to predict multiple output words**. 

### 1. Input Layer
* Takes the single target word as a one-hot vector ($x$).

### 2. Hidden Layer
* Size = $N$ (your desired embedding dimension).
* **No activation function** is used here.
* **No averaging** happens here (unlike CBOW), because there is only one input vector. It is a direct matrix multiplication (a simple lookup of the target word's embedding).

### 3. Output Layer
* Size = $V$ (the total vocabulary size).
* Uses a **softmax** activation function to output a probability distribution. The model tries to output high probabilities for *all* the actual context words.

---

## 6. Weight Matrices
Exactly like CBOW, the network relies on two weight matrices:
* **Input-to-Hidden matrix:** $W \in \mathbb{R}^{V \times N}$
* **Hidden-to-Output matrix:** $W' \in \mathbb{R}^{N \times V}$

**Important:** The matrix $W$ becomes the final word embeddings after training. The $W'$ matrix is discarded.

---

## 7. Mathematical Formulation (Forward Propagation)

### Step 1: Hidden Representation ($h$)
The hidden layer simply looks up the embedding for the single target word:

$$h = W^T x$$

* $x$ = one-hot vector of the target word.

### Step 2: Output Scores ($u$)
The hidden representation is multiplied by the second weight matrix to get raw scores for every word in the vocabulary:

$$u = (W')^T h$$

### Step 3: Softmax ($\hat{y}$)
Convert the raw scores into a probability distribution:

$$\hat{y}_j = \frac{e^{u_j}}{\sum_{k=1}^{V} e^{u_k}}$$

---

## 8. Loss Function & Backpropagation
Because Skip-gram predicts *multiple* context words (let's say there are $C$ context words in our window), the total loss is the sum of the Cross-Entropy Loss for each context word:

$$E = -\sum_{c=1}^{C} \log(\hat{y}_{c, \text{actual}})$$

Weights are updated using gradient descent to minimize this combined loss:

$$W_{\text{new}} = W_{\text{old}} - \eta \frac{\partial E}{\partial W}$$

### Pro-Tip: Optimization (Negative Sampling)
Because computing the softmax denominator over a massive vocabulary ($V$) for every single context word is extremely computationally expensive, real-world Skip-gram models use **Negative Sampling**. Instead of updating weights for all $V$ words, the model only updates the weights for the *true* context word and a small handful (2 to 10) of randomly selected "negative" (incorrect) words.

---

## 9. Training Process
1. Slide the context window across the entire corpus.
2. Generate all (target, context) training pairs.
3. Perform a forward pass using the target word.
4. Compute the loss across the true context words.
5. Backpropagate the error.
6. Update weights $W$ and $W'$.
7. Repeat for the entire corpus.

---

## 10. Advantages & Disadvantages

### Advantages
* **Rare Words:** Performs **exceptionally well with rare words**. Because it treats every (target, context) pair as a separate training observation, rare words get more distinct updates compared to CBOW's averaging method.
* **Nuance:** Captures highly complex semantic relationships and finer details in the vector space.

### Disadvantages
* **Speed:** **Slower to train** than CBOW. It has to make multiple predictions (and calculate loss multiple times) for a single context window.
* **Data Requirements:** Tends to require **larger datasets** to fully realize its potential compared to CBOW.

---

* Skip-gram predicts **surrounding context words** using a single **target word**.
* Uses a **One-to-Many** architecture (one input $\rightarrow$ multiple outputs).
* The hidden layer performs a **direct lookup** (no averaging).
* The matrix $W$ contains the final word embeddings.
* Result: Highly detailed vector representations that excel at capturing rare words and complex semantic relationships.

## When to Use CBOW vs Skip-gram

Both **CBOW** and **Skip-gram** are Word2Vec models used to learn word embeddings, but they differ in how they learn from data and when they should be used.

---

## Basic Difference

* **CBOW** → Predicts target word from context
* **Skip-gram** → Predicts context words from target

---

## When to Use CBOW

CBOW is preferred when:

* You have a **smaller dataset**
* You want **faster training**
* You need **general word representations**
* Your data is relatively **clean and less noisy**

### Why?

* CBOW averages context → smooth learning
* Computationally efficient
* Less sensitive to rare words

---

## When to Use Skip-gram

Skip-gram is preferred when:

* You have a **large dataset**
* You want **better accuracy**
* You need to learn **rare word representations**
* You want to capture **fine-grained semantic relationships**

### Why?

* Learns from each (target → context) pair separately
* Handles rare words better
* Captures more detailed relationships

---

## Practical Comparison

| Feature             | CBOW                     | Skip-gram           |
| ------------------- | ------------------------ | ------------------- |
| Training Speed      | Faster                   | Slower              |
| Data Requirement    | Small dataset            | Large dataset       |
| Accuracy            | Moderate                 | Higher              |
| Rare Words Handling | Poor                     | Good                |
| Context Learning    | Averaged (less detailed) | Detailed            |
| Use Case            | General tasks            | Deep semantic tasks |

---

## Example Intuition

Sentence:

> "I love machine learning"

* **CBOW** → uses ["I", "love", "learning"] → predicts "machine"
* **Skip-gram** → uses "machine" → predicts ["I", "love", "learning"]

---

## Key Insight

* Use **CBOW** when speed and efficiency matter
* Use **Skip-gram** when accuracy and deeper understanding matter

---


* CBOW → fast, efficient, good for small data
* Skip-gram → slower, more accurate, better for large data and rare words
* Choice depends on **dataset size and problem complexity**


## How to Improve CBOW and Skip-gram Models

Both **CBOW** and **Skip-gram** can be improved by tuning hyperparameters, optimizing training techniques, and improving data quality.

---

## 1. Increase Training Data

* More data → better context learning
* Improves quality of embeddings
* Especially important for **Skip-gram**

---

## 2. Tune Hyperparameters

### (a) Embedding Dimension (Vector Size)

* Typical values: 100, 200, 300
* Higher dimension → more expressive embeddings
* Too high → overfitting + slower training

---

### (b) Window Size

* Controls context range
* Small window → syntactic relationships
* Large window → semantic relationships

---

### (c) Learning Rate

* Too high → unstable training
* Too low → slow convergence
* Use gradual decay

---

## 3. Use Negative Sampling

* Instead of predicting all words, sample a few negative examples
* Makes training much faster
* Improves performance

---

## 4. Use Hierarchical Softmax

* Efficient alternative to softmax
* Useful for large vocabularies

---

## 5. Subsampling Frequent Words

* Remove very frequent words (like "the", "is") probabilistically
* Reduces noise
* Speeds up training

---

## 6. Better Text Preprocessing

* Lowercasing
* Remove noise (punctuation, special characters)
* Stopword handling (carefully)
* Lemmatization or stemming

---

## 7. Use Larger Context Variety

* Train on diverse datasets
* Helps generalization

---

## 8. Train for More Epochs

* More iterations → better learning
* But too many → overfitting

---

## 9. Use Pretrained Embeddings

* Word2Vec (Google News)
* GloVe
* FastText

Fine-tune instead of training from scratch

---

## 10. Choose Model Based on Use Case

* CBOW → faster, good for small data
* Skip-gram → better for rare words and large data

---

## Key Insight

Improving Word2Vec models is mainly about:

* Better **data**
* Better **hyperparameters**
* Better **training techniques**

---


* Increase data + tune parameters
* Use **negative sampling + subsampling**
* Clean and preprocess text properly
* Choose CBOW or Skip-gram based on task
* Use pretrained models when possible
