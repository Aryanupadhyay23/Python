## TF-IDF (Term Frequency – Inverse Document Frequency)

**TF-IDF** is a text representation technique used in Natural Language Processing (NLP) to measure how important a word is in a document relative to a collection of documents (corpus).

It improves over the Bag of Words model by giving **less importance to common words** and **more importance to rare but meaningful words**.

---

### What is TF-IDF?

$$\text{TF-IDF} = \text{TF} \times \text{IDF}$$

*   **TF (Term Frequency):** How frequently a word appears in a document.
*   **IDF (Inverse Document Frequency):** How rare a word is across all documents.

---

### Term Frequency (TF)

$$TF(t,d)=\frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d}$$

*   **Higher TF** $\rightarrow$ The word is important in that specific document.

---

### Inverse Document Frequency (IDF)

$$IDF(t)=\log\left(\frac{N}{df(t)}\right)$$

Where:
*   $N$ = Total number of documents.
*   $df(t)$ = Number of documents containing term $t$.

*   **Higher IDF** $\rightarrow$ The word is rare across the entire collection of documents.

---

### Final TF-IDF Formula

$$\text{TF-IDF}(t,d)=TF(t,d)\times IDF(t)$$

---

### Example

Consider two sentences:
1. "I love machine learning"
2. "I love deep learning"

*   Words like **"I" and "love"** appear in both documents $\rightarrow$ **lower importance**.
*   Words like **"machine" and "deep"** appear in fewer documents $\rightarrow$ **higher importance**.

---

### Why Use TF-IDF?

*   Reduces the impact of **common words**.
*   Highlights **important words**.
*   Improves performance in:
    *   Text classification
    *   Search engines
    *   Information retrieval

---

### Advantages

*   Performs better than simple Bag of Words models.
*   Reduces noise from highly frequent words (like "the", "a", "is").
*   Effectively captures the relative importance of words.

---

### Disadvantages

*   Still ignores **word order** and context.
*   Provides no deep **semantic understanding** (e.g., cannot recognize that "car" and "automobile" mean the same thing).
*   Creates a highly **sparse representation** (lots of zeros) for large vocabularies.

---


*   TF-IDF measures the **importance of words**.
*   It strategically combines **frequency** (within a single document) with **rarity** (across all documents).
*   It is much more effective than simple frequency-based methods.
*   It remains a foundational and widely used tool in NLP and search systems.

```python
import nltk
import re
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download('stopwords')
```

```python
corpus = [
    "This product is really amazing and I am very happy with its performance",
    "I did not like the product it was bad and disappointing",
    "The quality of this item is excellent and works perfectly",
    "Very poor experience the product broke after one use",
    "Absolutely fantastic quality and great value for money",
    "The delivery was late but the product quality is good",
    "I am extremely satisfied with this purchase and recommend it",
    "Worst product ever do not buy this item",
    "The packaging was nice and the product works as expected",
    "I love this product it is worth every penny"
]
```

```python
stop_words = set(stopwords.words('english'))
ps = PorterStemmer()
```

```python
processed_corpus = []

for sentence in corpus:
    # Lowercase
    sentence = sentence.lower()
    
    # Remove non-alphabet characters
    sentence = re.sub('[^a-z]', ' ', sentence)
    
    # Tokenization
    words = sentence.split()
    
    # Remove stopwords + stemming
    words = [ps.stem(word) for word in words if word not in stop_words]
    
    # Join back
    sentence = ' '.join(words)
    
    processed_corpus.append(sentence)

processed_corpus
```

```python
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(processed_corpus)
```

```python
print(tfidf.get_feature_names_out())
```

```python
print(X.toarray())
```

```python
tfidf_df = pd.DataFrame(X.toarray(), columns=tfidf.get_feature_names_out())
tfidf_df
```

```python
tfidf = TfidfVectorizer(ngram_range=(2,2))
X = tfidf.fit_transform(processed_corpus)
```
