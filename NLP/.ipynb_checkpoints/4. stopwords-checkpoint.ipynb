## Stopwords Removal

### What are Stopwords?

**Stopwords** are common words in a language that usually do not carry significant meaning in text analysis.

**Examples:**
is, the, in, and, of, to, a, an

---

### Why Remove Stopwords?

* Reduce **noise** in text
* Decrease **dimensionality**
* Improve **model performance**
* Focus on **important words**

---

### Example

**Original Sentence:**

> This product is really amazing and I am very happy with its performance.

**After Stopwords Removal:**

> product really amazing happy performance

---

### Code Example

```
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text = "This product is really amazing and I am very happy with its performance"

stop_words = set(stopwords.words('english'))

words = word_tokenize(text)

filtered_words = [word for word in words if word.lower() not in stop_words]

print(filtered_words)
```

---

* Stopwords are **high-frequency, low-meaning words**
* Removing them helps focus on **useful information**
* Should be used carefully (e.g., *not* can be important in sentiment analysis)


````python
paragraph = """Machine learning is a powerful field of artificial intelligence that enables systems to learn patterns from data and make predictions or decisions without being explicitly programmed. 
It is widely used in real-world applications such as recommendation systems, fraud detection, healthcare diagnostics, and self-driving cars. 
By leveraging algorithms and statistical techniques, machine learning models improve their performance over time as they are exposed to more data. 
Understanding the fundamentals of data preprocessing, model selection, and evaluation is essential for building effective and reliable machine learning solutions."""
````

````python
from nltk.stem import PorterStemmer
from nltk.stem import SnowballStemmer
````

````python
from nltk.corpus import stopwords
````

````python
import nltk
nltk.download('stopwords')
````

````python
stopwords.words('english')
````

````python
stemmer = PorterStemmer()
````

````python
sentences = nltk.sent_tokenize(paragraph)
````

````python
type(sentences)
````

````python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)
````

````python
print(sentences)
````

````python
snowballstemmer = SnowballStemmer('english')
````

````python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [snowballstemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)
````

````python
print(sentences)
````

````python
from nltk.stem import WordNetLemmatizer
````

````python
lemmatizer = WordNetLemmatizer()
````

````python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [lemmatizer.lemmatize(word, pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)
````

````python
print(sentences)
````
