## WordNet Lemmatizer

**Lemmatization** is a text preprocessing technique similar to stemming, but more advanced and meaningful.

The output of lemmatization is called a **lemma**, which is a **valid root word** (unlike stemming, which may produce invalid words).

---

## Key Difference from Stemming

* **Stemming** → May produce non-meaningful root (e.g., *history → histori*)
* **Lemmatization** → Produces a valid dictionary word (e.g., *history → history*)

---

## What is WordNet Lemmatizer?

In NLTK, the **WordNetLemmatizer** is used for lemmatization.

* It is a wrapper around the **WordNet corpus**
* Uses the **morphy()** function internally
* Converts words into their **base (dictionary) form**

---

## Example

```
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["running", "better", "wolves", "feet"]

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

---

## Output

```
running -> running  
better  -> better  
wolves  -> wolf  
feet    -> foot  
```

---

## Important Note (POS Tagging)

By default, lemmatizer assumes **noun (pos='n')**, so results may not always be correct.

To get better results, specify **Part of Speech (POS)**:

```
from nltk.corpus import wordnet

print(lemmatizer.lemmatize("running", pos=wordnet.VERB))  # run
print(lemmatizer.lemmatize("better", pos=wordnet.ADJ))   # good
```

---

* Produces **meaningful root words**
* Uses vocabulary and grammar (WordNet)
* More accurate than stemming
* Requires POS tagging for best results
* Lemmatization → converts word to **valid root form (lemma)**
* WordNetLemmatizer uses **WordNet corpus + morphy()**
* More accurate than stemming but slightly slower
* Best used when **meaning and correctness matter**


````python
from nltk.stem import WordNetLemmatizer
````

````python
lemmatizer = WordNetLemmatizer()
````

## POS Tags for WordNet Lemmatizer

**POS (Part of Speech)** tags are used to specify the grammatical role of a word.
Providing POS improves the accuracy of lemmatization.

### Common POS Tags

```
n → Noun  
v → Verb  
a → Adjective  
r → Adverb  
```

---

### Example

```
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("running", pos='v'))  # run
print(lemmatizer.lemmatize("better", pos='a'))   # good
```

---

* Without POS → default is **noun (n)**
* With POS → more accurate lemmatization

````python
lemmatizer.lemmatize("going", pos='v')
````

````python
words = [
    "eating", "eaten", "eats",
    "running", "runs", "ran",
    "playing", "played", "plays",
    "studies", "studying", "studied",
    "fairly", "easily", "happiness",
    "goes", "going", "gone"
]
````

````python
for word in words:
    print(word + " ---> " + lemmatizer.lemmatize(word, pos='v'))
````

````python
lemmatizer.lemmatize("goes")
````
