## creating strings

```python
s = 'hello'
print(s)
```

```python
s = "hello"
print(s)
```

```python
s = '''hello'''
print(s)
```

```python
s = "it's raining outside"
print(s)
```

```python
s = """
hello
how are you
"""
print(s)
```

## accessing substrings from a string

```python
s = "hello world"
for i in s:
    print(i, end = " ")
```

```python
print(s[4])
```

```python
print(s[8])
```

```python
print(s[-2])
```

```python
print(s[2:9:2])
```

```python
print(s[2:9:1])
```

```python
print(s[9:2:-2])
```

```python
print(s[9:2:-1])
```

```python
print(s[::-1])
```

```python
print(s[-1:-6:-1])
```

## Editing and Deleting in Strings

```python
# strings are immutable

s = "hello world"
s[0] = "H"
print(s)
```

```python
s = "hello world"
del s
print(s)
```

```python
s = "hello world"
del s[-1:-5:2]
print(s)
```
