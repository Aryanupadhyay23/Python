```python
number = int(input("Enter a number: "))
sum = 0
count = 1
while count <= number:
    sum += count
    count += 1
print("The sum of numbers from 1 to", number, "is:", sum)
```

```python
number = 5
sum = 0
for i in range(1, number + 1):
    sum += i
print("The sum of numbers from 1 to", number, "is:", sum)
```
