## if-else

```python
email = input("enter email : ")
password = input("enter password : ")

if email == "aryan@gmail.com" and password == '1234':
    print("login")
else:
    print("logout")
```

```python
email = input("enter email : ")
password = input("enter password : ")

if email == "aryan@gmail.com" and password == '1234':
    print("login")
    
elif email == "aryan@gmail.com" and password != '1234':
    print("print incorrect password")
    password = input("enter password : ")
    if password == "1234":
        print("login")
    else:
        print("wrong password")
else:
    print("wrong credentials")
```

```python
a = int(input("first number : "))
b = int(input("first number : "))
c = int(input("first number : "))

if a < b and a < c:
    print("smallest : ", a)
elif b < c:
    print("smallest : ", b)
else:
    print("smallest : ", c)
```

## Loops

```python
number = int(input("enter number : "))
i = 1
while i < 11:
    print(number, "*", i, "=", number* i)
    i+=1
```

```python
x = 1
while x < 3:
    print(x)
    x += 1
else:
    print("limit crossed")
```

```python
import random

jackpot = random.randint(1, 100)
guess = int(input("guess the number : "))

counter = 1

while guess != jackpot:
    if guess < jackpot:
        print("guess higher")
    else:
        print("guess lower")

    guess = int(input("guess again : "))
    counter += 1

else:
    print("correct guess")
    print("attempts : ", counter)
```

## Nested Loops

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

```python
rows = int(input("enter number of rows : "))
for i in range(1, rows+1):
    for j in range(1, i+1):
        print("*", end='')
    print()
```

```python
rows = int(input("enter number of rows : "))
for i in range(1, rows+1):
    for j in range(1, i+1):
        print(j, end='')
    for k in range(i-1, 0, -1):
        print(k, end='')
    print()
```

### Loop Control Statement
- break
- continue
- pass

```python
for i in range(1,10):
    if i == 5:
        break
    print(i)
```

```python
lower = int(input("enter lower range : "))
upper = int(input("enter upper range : "))

for i in range(lower, upper+1):
    for j in range(2, i):
        if i % j == 0:
            break
    else:
        print(i)
```

```python
for i in range(1, 10):
    if i == 5:
        continue
    print(i)
```

```python
for i in range(1, 10):
    pass
```
